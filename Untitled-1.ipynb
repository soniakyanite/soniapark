```python
import yfinance as yf
import matplotlib.pyplot as plt

# ~200 trading days; pull extra calendar days, then keep the last 200 rows
df = yf.download("^GSPC", period="1y", auto_adjust=True, progress=False)
close = df["Close"].squeeze().dropna().tail(200)

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(close.index, close.values, color="#2a6fdb", linewidth=2)
ax.set_title("S&P 500 close, last 200 trading days", loc="left", fontsize=13)
ax.set_ylabel("Index level")
ax.grid(axis="y", color="#e5e5e5", linewidth=0.8)
ax.set_axisbelow(True)
for side in ("top", "right", "left"):
    ax.spines[side].set_visible(False)
ax.tick_params(length=0)
ax.annotate(f"{close.iloc[-1]:,.0f}", (close.index[-1], close.iloc[-1]),
            xytext=(6, 0), textcoords="offset points", va="center")
plt.tight_layout()
plt.show()
```
